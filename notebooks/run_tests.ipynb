```python
import shutil
import pytest

for name in ("test_bronze.py", "test_gold.py", "test_cohorts_monitoring.py"):
    source_path = f"../tests/{name}"
    temp_path = f"/tmp/{name}"
    shutil.copy(source_path, temp_path)

result = pytest.main([
    "/tmp/test_bronze.py",
    "/tmp/test_gold.py",
    "/tmp/test_cohorts_monitoring.py",
    "-v",
    "-p", "no:cacheprovider",
])

print("pytest exit code:", result)
```
