```python
from pyspark.sql import SparkSession
from src.analysis.activation_rate import calculate_activation_rate

spark = SparkSession.getActiveSession()

result = calculate_activation_rate(spark, "BUILDING")

display(result)
```

```python
import matplotlib.pyplot as plt

row = result.collect()[0]

activated = row["activated_customers"]
total = row["total_customers"]
not_activated = total - activated

plt.figure(figsize=(6, 6))

plt.pie(
    [activated, not_activated],
    labels=["Activated", "Not activated"],
    autopct="%1.1f%%",
    startangle=90
)

plt.title("BUILDING Segment Activation Rate")
plt.axis("equal")

plt.show()
```
